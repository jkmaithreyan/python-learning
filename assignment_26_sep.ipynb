1. Problem Statement:

Your task: Write a Python function that returns the total sales amount for each city, counting only orders with a chosen status.

You are a data analyst at an online electronics store. Your manager uses a pivot table with Order status in the filter area, City in rows and Sales amount summed. You need the same group-by result in Python, so the check can run without opening a spreadsheet.

2. Approach:

Start an empty dictionary to hold one total per city.
Use a for loop to go through each order in the list.
Keep only the orders whose status matches the status parameter.
Add each kept order's sales amount to its city's total, starting a new entry the first time a city appears.
Return the dictionary from the function.
Use plain Python only, with no external libraries.

```python
orders = [
    {"order_id": "O101", "city": "Pune", "product": "Laptop", "sales_amount": 850, "status": "Delivered"},
    {"order_id": "O102", "city": "Delhi", "product": "Monitor", "sales_amount": 300, "status": "Cancelled"},
    {"order_id": "O103", "city": "Pune", "product": "Headphones", "sales_amount": 120, "status": "Delivered"},
    {"order_id": "O104", "city": "Bengaluru", "product": "Laptop", "sales_amount": 920, "status": "Delivered"},
    {"order_id": "O105", "city": "Delhi", "product": "Laptop", "sales_amount": 780, "status": "Delivered"},
    {"order_id": "O106", "city": "Bengaluru", "product": "Monitor", "sales_amount": 310, "status": "Pending"},
    {"order_id": "O107", "city": "Pune", "product": "Monitor", "sales_amount": 290, "status": "Cancelled"},
    {"order_id": "O108", "city": "Delhi", "product": "Headphones", "sales_amount": 150, "status": "Delivered"},
]
```

```python
def sales_by_city(orders, status):
    totals = {}

    for order in orders:
        if order["status"] == status:
            if order["city"] not in totals:
                totals[order["city"]] = order["sales_amount"]
            else:
                totals[order["city"]] += order["sales_amount"]

    return totals

print(sales_by_city(orders, "Delivered"))
print(sales_by_city(orders, "Cancelled"))
```
