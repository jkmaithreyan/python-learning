```python
import pandas as pd
sales = pd.DataFrame({"store": ["N", "S", "N", "E", "S", "N", "E"],
                      "product": ["tea", "tea", "cake", "tea", "pie", "cake", "tea"],
                      "units": [10, None, 4, 7, 5, 4, 7]})
```

```python
# Print the shape of sales and the number of missing values in units in one print call.
print(sales.shape, sales["units"].isnull().sum())
print(sales)
```

```python
# Remove the repeated rows (keeping the first occurrence) into clean and print its shape.
clean = sales.drop_duplicates()
print(clean.shape)
print(clean)
```

```python
# Fill the missing units in clean with the mean of the available units 
# and print that mean together with the filled value at row label 1 in one print call.
units_mean = clean["units"].mean()
clean["units"] = clean["units"].fillna(units_mean)
print(units_mean, clean.loc[1]["units"])
```

```python
# Group clean by store, sum units, and print the totals of stores N, S and E in one print call.
total = clean.groupby("store")["units"].sum()
print(total.iloc[0], total.iloc[1], total.iloc[2])
```

```python
# Sort the store totals in descending order and print the top store and its total in one print call.
# Then print how many rows of clean are for tea using value_counts().
sorted_total = total.sort_values(ascending=False)
print(sorted_total.index[0] ,sorted_total.iloc[0])
print(clean["product"].value_counts()["tea"])
```

```python
# Explain why the repeated rows are removed before the mean is computed,
# why the missing value is filled rather than dropped,
# and what the row count would have been if duplicates had been checked only on the store column.

#The repeated rows are removed before calculating mean. because, if mean is calculated before droping duplicates, repeated values will affect the correct mean value.
#if we drop the missing value, it will erase the information of the store and product data. so, we fill it with the mean value.
#There would be only 3 rows because all rows with the same store would be treated as duplicates.
```
