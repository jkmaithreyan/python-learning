1. Problem Statement:

Your task: Write a Python function that turns churn probabilities into labels at a given threshold and returns precision and recall.

You are a data scientist on the retention team of a food-delivery platform. A logistic regression model has already given ten delivery partners a churn probability. The team wants to see how precision and recall change at thresholds 0.3, 0.5 and 0.7 before picking one.

2. Approach:

Read each row as a partner ID, a churn probability, and the actual label (1 = churned, 0 = not churned).
Label a row 1 when its probability is at or above the threshold. Otherwise label it 0.
Count true positives, false positives and false negatives by comparing each predicted label with the actual label.
Compute precision as the share of predicted positives that are actually positive.
Compute recall as the share of actual positives the model catches. Return 0.0 for a metric whose denominator is zero.
Use plain Python only, with no external libraries. Print one line per threshold in the format shown below.

3. Expected Output:

Sample input (partner_id, churn_probability, churned):

P01, 0.82, 1
P02, 0.64, 1
P03, 0.45, 0
P04, 0.47, 1
P05, 0.38, 0
P06, 0.33, 1
P07, 0.28, 0
P08, 0.21, 0
P09, 0.12, 0
P10, 0.68, 0
Expected output for that input:

threshold=0.3 precision=0.57 recall=1.00
threshold=0.5 precision=0.67 recall=0.50
threshold=0.7 precision=1.00 recall=0.25

```python
# Sample data is already filled in for you: (partner_id, churn_probability, churned)
SAMPLE = [
    ("P01", 0.82, 1),
    ("P02", 0.64, 1),
    ("P03", 0.45, 0),
    ("P04", 0.47, 1),
    ("P05", 0.38, 0),
    ("P06", 0.33, 1),
    ("P07", 0.28, 0),
    ("P08", 0.21, 0),
    ("P09", 0.12, 0),
    ("P10", 0.68, 0),
]


def evaluate_threshold(rows, threshold):
    
    tp = 0
    fp = 0
    fn = 0


    for partner_id, probability, actual in rows:
        if probability >= threshold:
            predicted = 1
        else:
            predicted = 0

        if actual==1 and predicted==1:
            tp += 1
        elif actual==0 and predicted==1:
            fp += 1
        elif actual==1 and predicted==0:
            fn += 1

    if tp+fp == 0:
        precision = 0
    else:
        precision = round((tp / (tp+fp)), 2)

    if tp+fn == 0:
        recall = 0
    else:
        recall = round((tp / (tp+fn)), 2)

    return precision, recall



if __name__ == "__main__":
    results = []

    for t in [0.3, 0.5, 0.7]:
        precision, recall = evaluate_threshold(SAMPLE, t)
        print(f"threshold={t} precision={precision:.2f} recall={recall:.2f}")
        results.append((t, precision, recall))

    best = max(results, key=lambda x:x[2])

    print(f"Highest recall threshold: {best[0]}")
        
```
