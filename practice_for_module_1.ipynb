A retail analyst has the following sales data:

numbers = [1, 5, 4, 7, 9, 10, 55, 75, 33]

Write a function called analyze_sales(numbers) that:

Uses a for loop with range() and len() to access each value by its index.

Prints each value with its position, in the format Position 0: 1.

Calculates the total of all values.

Separately calculates the total of only the even values using %.

Returns both totals as a tuple: (total_sum, even_sum).

```python
numbers = [1, 5, 4, 7, 9, 10, 55, 75, 33]

def analyze_sales(numbers):
    total_sum = 0
    even_sum = 0

    for i in range(len(numbers)):
        print(f"Position {i}: {numbers[i]}")

        total_sum += numbers[i]

        if numbers[i] % 2 == 0:
            even_sum += numbers[i]

    return total_sum, even_sum

print(analyze_sales(numbers))    
```

Question 10 — Find Even Numbers
Coding challenge

You are given a list of numbers:

numbers = [12, 7, 20, 15, 8, 3, 10]

Write a function called get_even_numbers(numbers) that:

Creates an empty list called even_numbers.

Uses a for loop to iterate directly over the list.

Uses % to check whether each number is even.

Adds even numbers to even_numbers.

Returns the list.

```python
numbers = [12, 7, 20, 15, 8, 3, 10]

def get_even_numbers(numbers):

    even_numbers = []

    for num in numbers:
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers

print(get_even_numbers(numbers))
```

Coding Question 1/10 — Character Frequency
Write a function called count_character(word, character) that:
- Takes a string word
- Takes a specific character
- Counts how many times that character appears
- Returns the count

```python
def count_character(word, character):
    count = 0
    for char in word:
        if char.lower() == character.lower():
            count += 1
    return count

print(count_character("America", "A"))
```
